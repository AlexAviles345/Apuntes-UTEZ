```python
import pandas as pd
from sklearn.model_selection import train_test_split  # Dividir  en entrenamiento y prueba
from sklearn.preprocessing import StandardScaler  # Escalar las columnas a un rango especifico
from sklearn.neighbors import KNeighborsClassifier   # Clasificador 
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix  # Metricas 
# Accuracy, Presicion, F-Score
#         Falsos positivos - Cuantos valores que en realidad eran negativos, clasifico como positivos
#         Falsos negativos - Contrario
```

```python
df = pd.read_csv('iris.csv')
df["target"] = df["Species"].astype("category").cat.codes
df
```

```python
# Variables independientes
x = df[["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]]
# Target es la representación numerica de la especie, por eso es variable dependiente
y = df["target"]
```

```python
# Dividir los datos en entramiento y prueba
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)
#random_state genera los aleatorios de la misma forma tomando su numero (42) como base
```

```python
# Se reescala para que el algoritmo no piense que una columna es mas importante que otra
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train) # Esta escalando las 4 columnas en el rango entre -3 y 3
x_test_scaled = scaler.transform(x_test)
x_train_scaled
```

```python
knn = KNeighborsClassifier(n_neighbors=5) # Cuantos puntos cercanos a el va a considerar para hacer la clasificación
knn.fit(x_train_scaled, y_train) # Entrena y genera el modelo
```

```python
y_pred = knn.predict(x_test_scaled)  # Clasifica
```

```python
precision = accuracy_score(y_test, y_pred) # Cantidad de registros que clasifico correctamente
matriz_confusion = confusion_matrix(y_test, y_pred)
reporte = classification_report(y_test, y_pred, target_names=df["Species"].astype("category").cat.categories)
print(f"Precisión Global del Modelo: {precision * 100:.2f}%\n")
print("Matriz de Confusión:\n", matriz_confusion, "\n")
print("Reporte de Clasificación:\n", reporte)
```
