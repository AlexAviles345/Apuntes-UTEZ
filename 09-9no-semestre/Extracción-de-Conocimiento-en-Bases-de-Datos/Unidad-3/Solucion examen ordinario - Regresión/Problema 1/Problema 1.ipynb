## EN ESTA VERSION DEL EJERCICIO, LOS MISMOS DATOS DE ENTRENAMIENTO SE OCUPARON PARA HACER PRUEBAS. 

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

```python
datos = [
    [2450000, 1370000000],
    [2480000, 1350000000],
    [2540000, 1400000000],
    [2420000, 1330000000],
    [2350000, 1270000000],
    [2290000, 1210000000],
    [2400000, 1330000000],
    [2460000, 1350000000]
]
df_datos = pd.DataFrame(datos, columns=['cantidad_turistas', 'dinero gastado'])
df_datos
```

## Realiza los modelos lineal y polinomial de grado 2. De acuerdo con las métricas vistas en clase, ¿Cuál de los dos modelos se ajusta mejor a los datos?

```python
x_datos = df_datos[['cantidad_turistas']]
y_datos = df_datos[['dinero gastado']]
```

```python
# Modelos de grado 1 y 2
for i in range(1, 3):
    
    # Se define el gardo
    grado = i
    poly = PolynomialFeatures(degree=grado)
    x_poly = poly.fit_transform(x_datos)
    
    modelo_poly = LinearRegression()
    modelo_poly.fit(x_poly, y_datos)

    # Se realizan las prediciones
    predicciones = modelo_poly.predict(x_poly)

    # Calculo de RSME
    rmse_puntaje = np.sqrt(mean_squared_error(y_datos, predicciones))

    # Calculo de R2
    r2_puntaje = r2_score(y_datos, predicciones)
    
    print(f'predicciones de modelo grado {grado}: ', predicciones)
    print('esperados: ', y_datos)
    print(f'puntaje rsme de modelo grado {grado}: ', rmse_puntaje)
    print(f'puntaje r2 de modelo grado {grado}: ', r2_puntaje)
    print('----------------------------------')
```

```python
# RESPUESTA = El modelo de regresion polinomica de grado 1 se ajusta mejor a los datos en este caso, debido
# a que su r2 es la mas cercana a 1 (0.9257...) y su rsme es mas cercano a 0 (15288536.8...)
```
