Advertencia: Esto generará métricas engañosas (R² cercano a 1.0 y RMSE cercano a 0) debido al sobreajuste (overfitting), ya que el modelo simplemente está memorizando los datos en lugar de generalizar el patrón.
```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
```

```python
datos = [
    [3.5, 10, 7],
    [3.8, 15, 8],
    [2.9, 5, 6],
    [3.2, 8, 6],
    [3.9, 20, 7],
    [2.7, 3, 5],
    [3.6, 12, 8],
    [3.0, 6, 7]
]
df_datos = pd.DataFrame(datos, columns=['calificaciones', 'horas_estudio_semana', 'horas_estudio_noche'])
df_datos
```

## Realiza los modelos lineal y polinomial de grado 2, 3 y 4. De acuerdo con las métricas vistas en clase, ¿Cuál de los dos modelos se ajusta mejor a los datos?

```python
x_datos = df_datos[['horas_estudio_semana', 'horas_estudio_noche']]
y_datos = df_datos[['calificaciones']]
```

```python
# lineal
modelo_lineal = LinearRegression()
modelo_lineal.fit(x_datos, y_datos)
predicciones_lineal = modelo_lineal.predict(x_datos)
print('predicciones_lineal: ', predicciones_lineal)
print('esperados: ', y_datos)
```

```python
# polinomial grado 2
poly2 = PolynomialFeatures(degree=2)
x_poly2 = poly2.fit_transform(x_datos)
modelo_poly2 = LinearRegression()
modelo_poly2.fit(x_poly2, y_datos)
predicciones_poli2 = modelo_poly2.predict(x_poly2)
print('predicciones_poli2: ', predicciones_poli2)
print('esperados: ', y_datos)
```

```python
# polinomial grado 3
poly3 = PolynomialFeatures(degree=3)
x_poly3 = poly3.fit_transform(x_datos)
modelo_poly3 = LinearRegression()
modelo_poly3.fit(x_poly3, y_datos)
predicciones_poli3 = modelo_poly3.predict(x_poly3)
print('predicciones_poli3: ', predicciones_poli3)
print('esperados: ', y_datos)
```

```python
# polinomial grado 4
poly4 = PolynomialFeatures(degree=4)
x_poly4 = poly4.fit_transform(x_datos)
modelo_poly4 = LinearRegression()
modelo_poly4.fit(x_poly4, y_datos)
predicciones_poli4 = modelo_poly4.predict(x_poly4)
print('predicciones_poli4: ', predicciones_poli4)
print('esperados: ', y_datos)
```

```python
# comparar por rsme
rmse_lineal = np.sqrt(mean_squared_error(y_datos, predicciones_lineal))
rmse_poli2 = np.sqrt(mean_squared_error(y_datos, predicciones_poli2))
rmse_poli3 = np.sqrt(mean_squared_error(y_datos, predicciones_poli3))
rmse_poli4 = np.sqrt(mean_squared_error(y_datos, predicciones_poli4))
print('puntaje rsme lineal: ', rmse_lineal)
print('puntaje rsme poli2: ', rmse_poli2)
print('puntaje rsme poli3: ', rmse_poli3)
print('puntaje rsme poli4: ', rmse_poli4)
```

```python
# Comparar por R2
r2_lineal = r2_score(y_datos, predicciones_lineal)
r2_poli2 = r2_score(y_datos, predicciones_poli2)
r2_poli3 = r2_score(y_datos, predicciones_poli3)
r2_poli4 = r2_score(y_datos, predicciones_poli4)
print('puntaje r2 lineal: ', r2_lineal)
print('puntaje r2 poli2: ', r2_poli2)
print('puntaje r2 poli3: ', r2_poli3)
print('puntaje r2 poli4: ', r2_poli4)
```

```python
# RESPUESTA = El modelo de regresion polinomica de grado 2 se ajusta mejor a los datos en este caso, debido
# a que su r2 es la mas cercana a 1 (0.9949...) y su rsme es mas cercano a 0 (0.0293...)
```
