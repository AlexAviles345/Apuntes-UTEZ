```python
# Importar librerias
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score
```

### 1. Carga el dataset Walmart_Sales.csv

```python
df = pd.read_csv('Walmart_Sales.csv')
```

### 2. Realiza la elección de variables para realizar una regresión lineal múltiple con Holiday_Flag y Temperature y una regresión  polinomial múltiple con Fuel_Pricey Temperature para calcular Weekly_Sales.

```python
# Si no se le pone copy, en el paso 3 aparece advertencia de que se esta reescalando los mismo objetos en memoria de df
x_lineal = df[['Holiday_Flag', 'Temperature']].copy()
y_lineal = df['Weekly_Sales'].copy()

x_poly = df[['Fuel_Price', 'Temperature']].copy()
y_poly = df['Weekly_Sales'].copy()
```

### 3. Normaliza las variables independientes para la regresión polinomial múltiple.

```python
scaler_minmax = MinMaxScaler()
x_poly[['Fuel_Price', 'Temperature']] = scaler_minmax.fit_transform(x_poly[['Fuel_Price', 'Temperature']])
```

### 4. Estandariza las variables independientes para la regresión lineal múltiple.

```python
scaler_std = StandardScaler()
x_lineal[['Holiday_Flag', 'Temperature']] = scaler_std.fit_transform(x_lineal[['Holiday_Flag', 'Temperature']])
```

### 5. Realiza la división de valores en 80% para entrenamiento y 20% para prueba con los datos de la regresión lineal múltiple.

```python
# Datos de la lineal
x_train_lineal, x_test_lineal, y_train_lineal, y_test_lineal = train_test_split(x_lineal, y_lineal, test_size=0.2, random_state=42)

# Datos de la polinomial
x_train_poly, x_test_poly, y_train_poly, y_test_poly = train_test_split(x_poly, y_poly, test_size=0.2, random_state=42)
```

### 6. Entrena el algoritmo.

```python
# Modelo lineal entrenado
grado = 1
lineal = PolynomialFeatures(degree=grado) # SIrve tambien para modelos lineales
x_train_lineal_fit = lineal.fit_transform(x_train_lineal)
x_test_lineal_fit = lineal.transform(x_test_lineal)

# Entrenamiento del modelo
modelo_lineal = LinearRegression()
modelo_lineal.fit(x_train_lineal_fit, y_train_lineal)

predicciones_lineal = modelo_lineal.predict(x_test_lineal_fit)

# Calculo de metricas
r2 = r2_score(y_test_lineal, predicciones_lineal)
rsme = np.sqrt(mean_squared_error(y_test_lineal, predicciones_lineal))

print(f'R2de modelo lineal: {r2}')
print(f'RSME de modelo lineal: {rsme}')
```

```python
# Modelo polinomial entrenado
grado = 2
poly = PolynomialFeatures(degree=grado)
x_train_poly_fit = poly.fit_transform(x_train_poly)
x_test_poly_fit = poly.transform(x_test_poly)

# Entrenamiento del modelo
modelo_poly = LinearRegression()
modelo_poly.fit(x_train_poly_fit, y_train_poly)

predicciones_polinomial = modelo_poly.predict(x_test_poly_fit)

# Calculo de metricas
r2 = r2_score(y_test_poly, predicciones_polinomial)
rsme = np.sqrt(mean_squared_error(y_test_poly, predicciones_polinomial))

print(f'R2de modelo polinomial: {r2}')
print(f'RSME de modelo polinomial: {rsme}')
```

### 7. Cuál sería el valor de ventas semanales si no es semana de festividad y la temperatura es de 60 grados farenheit?

```python
dato = pd.DataFrame([[0, 60.00]], columns=['Holiday_Flag', 'Temperature'])
# Se escalan los datos
dato_escalado = scaler_std.transform(dato)
# Se convierte en formato polinomial
dato_fit = lineal.transform(dato_escalado)
# Se hace la predicción
predicciones_lineal = modelo_lineal.predict(dato_fit)
predicciones_lineal
```
