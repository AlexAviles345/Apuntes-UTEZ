```python
import pandas as pd
```

```python
df_airbnb = pd.read_csv('AB_NYC_2019.csv')
```

```python
df_laptops = pd.read_excel('ventas_electronicas.xlsx', sheet_name='Laptops')
```

CONDICIONALES

```python
#Ventas de laptops de la región Norte
df_laptops[df_laptops['Region'] == 'Norte']
```

```python
# Ventas cuyo precio sea mayor a 600
df_laptops[df_laptops['Precio'] > 600]
```

```python
# Ventas cuyo descuento está entre el 20% y el 40%
df_laptops[(df_laptops['Descuento'] > 0.20) & (df_laptops['Descuento'] < 0.40)]
```

```python
#Ventas de laptops de la marca Xiaomi o Dell
df_laptops[(df_laptops['Marca']=='Xiaomi') | (df_laptops['Marca']=='Dell')]
```

```python
#Ventas de laptops de la marca Xiaomi o Dell
df_laptops[df_laptops['Marca'].isin(['Xiaomi', 'Dell'])]
```

PATRONES

```python
# Alojamientos cuyo nombre empieza con la palabra Clean
# Accesor - str convertir un dataframe en un StringFunctions
df_airbnb[df_airbnb['name'].str.startswith('Clean')]  # LIKE 'Clean%'
```

```python
# Alojamientos cuyo nombre contiene la palabra Home
df_airbnb[df_airbnb['name'].str.contains('Home')]  # LIKE '%Home%'
```

```python
# Alojamientos cuyo nombre contiene la palabra Home
# case que ignore las mayusculas
# na que solo analice los registros llenos e ignore los vacios
df_airbnb[df_airbnb['name'].str.contains('Home', case=False, na=False)]  # LIKE '%Home%'
```

FUNCIONES DE AGREGACION

```python
# AVG, COUNT, MAX, MIN, SUM funciones de agregación
# Cual es el precio maximo de los alojamientos?
df_airbnb['price'].max()
```

```python
# Promedio
df_airbnb['price'].mean()
```

```python
# Cuantos alojamientos tiene un precio mayor a 600
df_airbnb[df_airbnb['price'] > 600].count() # ERROR: ESTO DA POR CADA COLUMNA
```

```python
# Cuantos alojamientos tiene un precio mayor a 600 (COUNT)
(df_airbnb['price'] > 600).sum() 
```

```python
# Cuantos alojamientos tiene un precio mayor a 600 (COUNT)
df_airbnb[df_airbnb['price'] > 600].shape[0]
```

```python
# Cuantos alojamientos tiene un precio mayor a 600 (COUNT)
df_airbnb[df_airbnb['price'] > 600]['name'].count()
```

```python
# df[condicion][columnas]
df_airbnb[df_airbnb['price'] > 600][['name', 'price']]
```
